**PREFACE**

***Accuracy*** in model training is often a double edged sword, A layman would think that the higher the accuracy the better, but in reality \
it really isn't, as a matter of fact it could be a result that you would want to avoid, in some cases its the worst case scenario.

Imagine this scenario, two people are presented with the results of a model during training. One person would be a layman \
maybe the client that tasked you to make this model for his application. the second would be a Data Scientist. \
To keep it short they are told the Accuracy was 99.9% during training and testing. the layman would jump for joy and think the work is done \
and he can finally insert the model into his product however the scientist next to him would say that the work is far from finished. \

Why is this you ask? because in data science there is this concept of ***Overfitting***. 

imagine you would teach a child how to do math. in your tutoring sessions it seems that your student is doing pretty well. \
but when they arrived home after a quiz you find out they failed that quiz it turned out that after repeating the same practice \
questions over and over again they simply just memorized the questions and their corresponding answers, they didn't know what to do \
when presented with new questions, and hence they failed.

In the language of data science we call that "a model that failed to generalize". meaning that it simply memorized the dataset and returned a very high \
***Accuracy*** or a very low error. It failed to grasp the underlying patterns within the dataset hence failing in the real world test.

**A Quick Refresher on Neural Networks**

Unlike traditional machine learning models that often rely on simpler statistical formulas to find underlying patterns, \
neural networks use complex webs of interconnected neurons. They work by continuously adjusting the ***weights*** the strength \
of the connections on each neuron, not unlike a biological brain forming memories (hence the name) to make a prediction. \
In the grand scheme, their end goals are similar, but their internal organs are different. At least, they are different enough to warrant a preface for our main topic.



**What is L2 Regularization?**

L2 Regularization was an answer to an optimization problem statistics wherein numbers would explode in size, It works by using and Equitable approach to the numbers \
where it punishes small numbers not so much and punishing larger numbers the most, that way numbers are kept small.

it is expressed as this.

$$ \text{Total Loss} = \text{Data Loss} + \lambda \sum w^2 $$

The number that you want to keep from exploding is expressed as 'w' and squaring it gives it the power to punishes numbers disproportionaly. you can simply imagine: \
$$
\begin{aligned}
\text{Case 1:} \qquad & 0.1^2 = 0.01 \\
\text{Case 2:} \qquad & 10^2 = 100
\end{aligned}
$$

As you can see, L2 Regularization treats large numbers harshly, but it is much more lenient towards small numbers.


<br>

**Why is ~~Gamora~~ L2 Regularization?**

Now that we have prefaced What Overfitting is, What Neural Networks are, and what L2 Regularization is. We can finally cover why it can be used against overfitting. \
As mentioned before overfitting is the result of the model basically just memorizing the dataset and not being able to capture the underlying patterns in the dataset. \
In the case of Neural Networks this happens by the weights of neural networks going for an all or nothing approach of weight distribution. Take for example it is given cat pictures, \
it memorizes the general shapes of cats. four limbs, a torso, cute head, and a tail, the weights incharge of recognizing that are given way to much focus, \
resulting in only memorizing that specific pictures of cats, giving it pictures of cats in a loaf position, or curled up into a ball it won't be able to recognize it as a cat.

This is where L2 Regularizatiion steps in and basically tells the model to stop hoarding all the weight into one neuron, and encourages it to spread the weights more freely. \
It is able to do this due to L2 Regularization's ability to punish and discourage large numbers. by spreading out the weights more evenly it becomes more able to generalize.
